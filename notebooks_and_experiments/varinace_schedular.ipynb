```python
import torch
```

```python
betas = torch.linspace(start = 1e-4, end = 0.02, steps=1000)
```

```python
alphas = 1.0 - betas
alphas_cumprod = torch.cumprod(alphas, dim=0)
print(alphas_cumprod.shape, alphas_cumprod[0])
```

```python
 # alpha_bar_{t-1}, shifted by one, with 1.0 prepended for t=0
alphas_cumprod_prev = torch.cat([
            torch.tensor([1.0]), alphas_cumprod[:-1]
        ])
```

```python
def extract(arr, t, shape):
        # arr: precomputer lookup tables shape: (T, ) 
        # arr[0] is the value for timestep t = 0
        # t shape: (B, )
        # extract arr[t] and reshape
        out = arr.to(t.device)[t] #outputs arr[t] #output sqrt_alpha_cumprod[t]
        return out.reshape(t.shape[0], *([1] * (len(shape) - 1))) 
```

```python
betas  = torch.linspace(1e-3, 0.02, 1000)
```

```python
t = torch.randint(1, 1000, (64,)) 
```

```python
out = extract(betas, t, t.shape)
```
