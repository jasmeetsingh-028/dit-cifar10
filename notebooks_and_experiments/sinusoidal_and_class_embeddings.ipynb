```python
y = torch.randint(0, 10, (4,))
```

```python
y.shape
```

```python
dropout_prob = 0.5
null_token = 10
```

```python
drop_mask = torch.rand(y.shape[0], device=y.device) < dropout_prob
```

```python
drop_mask 
```

```python
y = torch.where(drop_mask, torch.full_like(y, null_token), y)  # overlay mask on y, put null on y indexes where mask is True
```
